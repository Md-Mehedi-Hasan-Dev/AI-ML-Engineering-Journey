### Importing Seaborn

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```


---

## Line Plot using Seaborn

### Loading Data from CSV file in Pandas DataFrame

```python
student = pd.read_csv('sns_data.csv')

student.head()
```

### Line Plot using `sns.lineplot(DataFrame, x-axis, y-axis)` & `errorbar` parameter

### Axes Level Plot -> single plot/graph

```python
sns.lineplot(data = student, x = 'week', y = 'attendance_rate', errorbar = None) # errorbar to remove the errorbar
```

#### Gender-wise Line Plot Separation using `hue` parameter

```python
sns.lineplot(data = student, x = 'week', y = 'attendance_rate', errorbar = None, hue = 'gender')
```

### Figure Level Plot -> can draw multiple axes in a figure

### Figure Level Ploting using `sns.relplot(kind, data)` -> can draw multiple plots of different kind

```python
# week vs attendance_rate
sns.relplot(kind = 'line', data = student, x = 'week', y = 'attendance_rate', errorbar = None, hue = 'gender')
```

```python
# week vs test_score
sns.relplot(kind = 'line', data = student, x = 'week', y = 'test_score', errorbar = None, hue = 'gender')
```

#### Class-level categorical line plot using `hue` parameter in `relplot()`

```python
# week vs test_score
sns.relplot(kind = 'line', data = student, x = 'week', y = 'test_score', errorbar = None, hue = 'class_level')
```

```python
# week vs attendance_rate
sns.relplot(kind = 'line', data = student, x = 'week', y = 'attendance_rate', errorbar = None, hue = 'class_level')
```

### Positive Correlation

```python
sns.scatterplot(data = student, x = 'study_hours', y = 'test_score')
```

#### Gender-wise Scatter Plot Separation using `hue` parameter

```python
sns.relplot(kind = 'scatter', data = student, x = 'study_hours', y = 'test_score', hue = 'gender', style = 'class_level')
```

## Facet Plot using Seaborn using `sns.relplot()`

### Loading Dataset from CSV to Pandas DataFrame

```python
student = pd.read_csv('student_dataset_complete.csv')

student.head()
```

### Scatter Plot using `sns.scatterplot()`

```python
sns.scatterplot(data = student, x = 'study_hours', y = 'test_score', hue = 'gender')
```

### Figure-level Scatter Plot using `sns.relplot()`

```python
sns.relplot(kind = 'scatter', data = student, x = 'study_hours', y = 'test_score', hue = 'gender')
```

### Facet Plot & Categorical wise axes split using `sns.relplot()` & `col` parameter

#### Col-wise axes Separation

```python
# col 'gender' wise axes separation
sns.relplot(kind = 'scatter', data = student, x = 'study_hours', y = 'test_score', col = 'gender')
```

```python
# col 'week' wise axes separation
sns.relplot(kind = 'scatter', data = student, x = 'study_hours', y = 'test_score', col = 'week', col_wrap = 2)
```

#### Row & Col wise axes separation

```python
# row 'hostel' & col 'gender' wise 'study_hours vs test_score' Figure-Level Scatter Plot -> Facet Plot
sns.relplot(kind = 'scatter', data = student, x = 'study_hours', y = 'test_score', col = 'gender', row = 'hostel')
```

## Bar Plot using Seaborn

### Bar Plot using `sns.barplot()` -> Categorical Value with Aggregate functions [x -> Category & y -> Aggregate Function]

#### y-axis aggregate fuction is mean (`estimator = 'mean'`, default)

```python
sns.barplot(data = student, x = 'gender', y = 'Marks', errorbar = None)
```

#### y-axis aggregate fuction is median (`estimator = 'median'`)

```python
sns.barplot(data = student, x = 'gender', y = 'Marks', errorbar = None, estimator = 'median') # estimator = np.median
```

#### y-axis aggregate fuction is max (`estimator = 'max'`)

```python
sns.barplot(data = student, x = 'gender', y = 'Marks', errorbar = None, estimator = 'max') # estimator = np.max
```

## Pair Plot & Joint Plot using Seaborn

### Pair Plot using `sns.pairplot()`

#### Col selection from DataFrame

```python
student_marks = student[['Marks', 'study_hours', 'attendance_rate', 'gender']]
student_marks
```

#### Pair plots using scatter plots (`kind = 'scatter'`, by default)

```python
sns.pairplot(data = student_marks, hue = 'gender')
```

#### Pair plots using histogram plots (`kind = 'hist'`)

```python
sns.pairplot(data = student_marks, kind = 'hist', hue = 'gender')
```

### Joint Plot using `sns.jointplot()`

```python
sns.jointplot(data = student_marks, x = 'study_hours', y = 'Marks')
```

#### Adding parameter `hue`

```python
sns.jointplot(data = student_marks, x = 'study_hours', y = 'Marks', hue = 'gender')
```

#### Adding parameter `kind`

```python
sns.jointplot(data = student_marks, x = 'study_hours', y = 'Marks', kind = 'kde')
```

#### Add both `hue` & `kind` parameter

```python
sns.jointplot(data = student_marks, x = 'study_hours', y = 'Marks', hue = 'gender', kind = 'kde')
```
