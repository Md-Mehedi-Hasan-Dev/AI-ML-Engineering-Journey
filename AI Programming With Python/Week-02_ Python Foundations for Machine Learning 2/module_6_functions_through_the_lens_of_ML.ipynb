## Different Type of Arguments

### Positional Arguments

```python
# Position of parameters and arguments matters
def adder(a, b, c):
  print(f"{a}, {b}, {c}")

adder(10, 20, 30)
```

```python
# variable sized arguments/parameters -> using args
def adder(*args):
  print(args)

adder(10, 20, 30, 40) # returns a tuple
```

```python
def adder(*args):
  for i in args:
    print(i)

adder(10, 20, 30, 40)
```

### Keyword Arguments

```python
def adder(a = 1, b = 2, c = 3): # assing default value
  print(f"{a}, {b}, {c}")

adder(a = 10, c = 30, b = 20) # assigning values to parament in the function call arguments -> keyword argument
```

```python
# variable sized arguments for keyword arguments -> using kwargs
def adder(**kwargs):
  print(kwargs)

adder(a = 10, b = 30, c = 40, n_estimator = 50) # returns a dictionary
```

```python
def adder(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}: {value}")

adder(a = 10, b = 30, c = 40, n_estimator = 50)
```

## Return Type

```python
# No Return
def welcome_message():
  print("Hello World!")

print(type(welcome_message()))
```

```python
# int return
def adder(a, b):
  return a + b

print(type(adder(10, 20)))
```

```python
# float return
def adder(a, b):
  return a + b

print(type(adder(10.5, 20)))
```

```python
# string return
def welcome_message(user = "None"):
  return f"hello {user}"

print(type(welcome_message()))
```

```python
# returning multiple values as tuple
def adder(a, b):
  return a, b

print(type(adder(10, 20)))

# accessing (unzip) tuple values returned from function
x, y = adder(10.5, 20.8)

print(x, y)
```

```python
# return as list, set & dictionary
def adder(a, b):
  return [a, b] # we can return anything as return type

print(type(adder(10, 20)))
```

## Generator/Yield

```python
lst = [x for x in range(1, 501)] # dataset

def data_loader(chunck_size, lst):
  for i in range(0, len(lst), chunck_size):
    yield lst[i : i + chunck_size]

x = data_loader(5, lst)

print(type(x))

print(x)
```

```python
print(next(x))
print(next(x))
print(next(x))
```

## Lambda, Map and Apply function

```python
import pandas as pd

df = pd.DataFrame({'temperature (F)': [32, 212, 98.6, 68]}) # converts the list to a dataframe and creates new column as "temperature (F)"
df # prints dataframe (only works in nootbook -> use print() for compiler)

df['temperature (C)'] = df['temperature (F)'].apply(lambda x: (x - 32) * 5/9) #creates new column and apply() converts (Fahrenheit) to (Celcius)
df # prints dataframe
```

```python
import pandas as pd

df = pd.DataFrame({'A': [1, 2, 3, 4], 'B': [5, 6, 7, 8]})

df

df['A'] = df['A'].map(lambda x: x ** 0.5) # map() -> applies function element-wise

df
```

```python
import pandas as pd

df = pd.DataFrame({'total_spent': [120, 600, 450, 800]})

# Creating a 'High Value' feature on the fly
df['is_high_value'] = df['total_spent'].apply(lambda x: 1 if x > 500 else 0) # apply() -> applies function row & column wise

df
```
