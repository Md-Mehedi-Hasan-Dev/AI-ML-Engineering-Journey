## String Methods

```python
string = "Welcome to Phitrom ML course"

# Length
print(len(string))
```

```python
string = "Welcome to Phitron ML course"

# find something in a string
print("Phitron" in string)

print("To" in string)
```

```python
string = "Welcome to Phitron ML course"

processed_string = string.upper()

print(processed_string)

# find something in a string
print("phitron" in processed_string)

print("to" in processed_string)
```

```python
string = "Welcome to Phitron ML course, Phitron"

processed_string = string.lower()

print(processed_string)

# find something in a string
print("phitron" in processed_string)

print("to" in processed_string)

# type of finding something in a string
print(type("to" in processed_string))
```

```python
# from beginning finding index of the word we're searching
start_index = processed_string.find("phitron")

# from beginning finding index of the word we're searching
end_index = processed_string.rfind("phitron")

print(start_index)

print(end_index)
```

```python
# counts the number of substring present in a string
count1 = processed_string.count("phitron")

count2 = processed_string.count("welcome")

print(count1)

print(count2)
```

```python
new_string = "Welcome to Phitron ML course, Phitron ML, Hello ML"

proc_str = new_string.lower()

# Replace a substring with a different string
rep_str = new_string.replace("ML", "AI/ML")

print(rep_str)
```

## Splitting, Joining & Formatting String

```python
prompt = "what is Phitron ?"

# spliting/tokenization of a string
tokens = prompt.split() # returns list

print(tokens)
```

```python
prompt = "what-is-Phitron-?"

prompt2 = "what,is,Phitron-?"

tokens = prompt.split("-")

tokens2 = prompt2.split(",")

print(tokens)

print(tokens2)
```

```python
# Joining tokens to form a string
sentence = "-".join(tokens)

print(type(tokens)) # list
print(type(tokens[0])) # tokens are str

print(sentence)

print(type(sentence))
```

```python
# Formatted String
name = "Naim"
age = 23
height = 5.5523

print(f"His name is {name.upper()}. He is {age}. He height is {height:.2}\n")
```

```python
info = f"His name is {name.upper()}. He is {age}. He height is {height:.3}"

print(info)

print(type(info))
```
