## Class Methods

```python
# defining class
class Phone: # class naming convention -> class name should be capitalized word and variable name starts with small letters
  category = "Electronics" # class attributes

  # constructor
  def __init__(self, model, battery, camera, battery_percentage = 100): # instance/obj attributes
    self.model = model
    self.battery = battery
    self.camera = camera
    self.battery_percentage = battery_percentage

  # methods
  def charge(self, hour):
    print(f"charge completed by {hour}")

  def capture(self, photo):
    if (self.battery_percentage) <= 0:
      print("No Charge")
    else:
      self.battery_percentage -= photo
      print(f"Photo Captured in {self.model}")
```

```python
# creating obj
apple = Phone("Iphone17", 3000, 40)
blackberry = Phone("BB17", 4000, 30)
motorola = Phone("M17", 3500, 35)

print(apple.battery_percentage)

apple.capture(10)
print(apple.battery_percentage)
```

## Inheritance

### Single Inheritance

```python
# child/derived class
class SmartPhone(Phone):
  def __init__(self, model, battery, camera, processor):
    super().__init__(model, battery, camera) # calling parent class constructor -> no self keyword while using super()
    self.processor = processor

  def charge(self, hour):
    print(f"fast charging in process")
    super().charge(hour) # calling parent class method -> no self keyword while using super()


class Cooling_Mechanism:
  def __init__(self, cooling_method):
    self.cooling_method = cooling_method

  # method
  def cooling_on(self):
    print(f"the system is being cool by {self.cooling_method}")
```

```python
pro = SmartPhone("X", 5000, 100, "SnapDragon")

print(pro.model)

pro.charge(1)
```

### Multiple Inheritance

```python
class SmartPhone_CoolingMode(SmartPhone, Cooling_Mechanism):
  def __init__(self, model, battery, camera, processor, cooling_method):
    SmartPhone.__init__(self, model, battery, camera, processor)
    Cooling_Mechanism.__init__(self, cooling_method)
```

```python
pro_cooling = SmartPhone_CoolingMode("T", 5000, 100, "SD", "Nitrogen")

print(pro_cooling.processor) # inherited from SmartPhone class
print(pro_cooling.battery) # inherited from Phone class
print(pro_cooling.cooling_method) # inherited from Cooling_Mechanism class
pro_cooling.cooling_on() # method of Cooling_Mechanism class
pro_cooling.charge(1) # method of SmartPhone class
```

## Polymorphism

### Method Overriding

```python
class Camera:
  def __init__(self, name):
    self.name = name

  # method
  def capture(self):
    print("a photo is captured")
```

```python
class Smart(Camera):
  def __init__(self, name, resolution):
    super().__init__(name)
    self.resolution = resolution

  # method overriding
  def capture(self):
    print("Photo is captured by Phone")

class DSLR(Camera):
  def __init__(self, name, resolution):
    super().__init__(name)
    self.resolution = resolution

  # method overriding
  def capture(self):
    print("Photo is captured by DSLR")

class Drone(Camera):
  def __init__(self, name, resolution):
    super().__init__(name)
    self.resolution = resolution

  # method overriding
  def capture(self):
    print("Photo is captured by Drone")

phone = Smart("Phone", 30)
dslr = DSLR("DSLR", 200)
drone = Drone("Drone", 150)

phone.capture()
dslr.capture()
drone.capture()
```

## Encapsulation

```python
# class section:
class Mobile:
  def __init__(self, name, model, imei):
    self.__name = name
    self.__model = model
    self.__imei = imei # private

  # method
  def charge(self):
    print("phone is charging")

  # to access imei value (just for reading)
  def imei_getter(self):
    return self.__imei

  # to access model value (just for reading)
  def model_getter(self):
    return self.__model

  # to access name value (just for reading)
  def name_getter(self):
    return self.__name

  # to change name value (writing)
  def name_setter(self, name):
    self.__name = name

# object section:
iphone = Mobile("Phone", "17", "1asdf")

print(iphone.name_getter())
print(iphone.model_getter())
print(iphone.imei_getter())

iphone.name_setter("Mehedi")
print(iphone.name_getter())

iphone.imei = "123456" # can't changed it because it is private
# print(iphone.imei) # can't access this way
print(iphone.imei_getter())
```

## Abstraction

```python
from abc import ABC, abstractmethod # importing abstract base class (ABC)

# abstract class
class Telephone(ABC): # ABC -> abstract base class
  @abstractmethod
  def make_call(self):
    pass # abstract methods will always be empty

  @abstractmethod
  def capture_photo(self):
    pass # abstract methods will always be empty

class Sphone(Telephone):
  def make_call(self):
    print("Making a call using SPhone")

  def capture_photo(self):
    print("Capured photo using SPhone")

class Iphone(Telephone):
  def make_call(self):
    print("Making a call using IPhone")

  def capture_photo(self):
    print("Capured photo using IPhone")
```

```python
# object section:
ip = Iphone()
sp = Sphone()

ip.make_call()
sp.make_call()

print()

ip.capture_photo()
sp.capture_photo()
```
