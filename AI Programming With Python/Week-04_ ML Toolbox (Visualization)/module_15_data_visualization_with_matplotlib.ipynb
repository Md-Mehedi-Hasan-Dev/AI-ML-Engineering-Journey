### Importing Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```


---

## Plotting 2D Line Plot using `plt.plot(x-axis, y-axis)`

```python
# plotting a students study hours of a week
hours = [2, 3, 4, 1, 5, 7, 3]
days = [1, 2, 3, 4, 5, 6, 7]

plt.plot(days, hours)
```

### Title of graph, x & y axis labels using `plt.title()`, `plt.xlabel()` & `plt.ylabel()`

```python
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Trend of a student study time over a week')

plt.plot(days, hours)
```

### Plotting multiple line plot on a single 2D plane

```python
# plotting two students study hours of a week
hours_1 = [2, 3, 4, 1, 5, 7, 3]
hours_2 = [1, 4, 3, 5, 7, 2, 1]
days = [1, 2, 3, 4, 5, 6, 7]

plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Trend of a student study time over a week')

plt.plot(days, hours_1)
plt.plot(days, hours_2)
```

## 2D Line Plot from Pandas DataFrame

```python
df = pd.read_csv('enrollment_data.csv')

df
```

### 2D Line Plotting of the DF using `.plot()` & labeling x & y axis and graph using `.xlabel()`, `ylabel()` & `.title()`

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df['Programming'])
```

### Plotting multiple 2D Line Plot in a single graph and labeling the line plot using `plt.legend()` & `label` parameter of `plt.plot()`

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df['Programming'], label = 'Programming')
plt.plot(df['Year'], df['Digital Marketing'], label = 'Digital Marketing')

plt.legend() # shows the labels in plt.plot()
```

### Changing the `loc` parameter of `.legend()` to move the legends location

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df['Programming'], label = 'Programming')
plt.plot(df['Year'], df['Digital Marketing'], label = 'Digital Marketing')

plt.legend(loc = 'lower right') # by default loc is set to 'best'
```


---

## Customizing 2D Line Plot using parameters -> `color`, `linestyle`, `marker` & Grid using `plt.grid()`

### Customizing 2D Line Plot using parameters `color`, `linewidth`, `linestyle`, `marker` & `markersize`

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df['Programming'], label = 'Programming', color = '#11b86f', linewidth = 3.5, linestyle = 'dashed', marker = 'o', markersize = 7)
plt.plot(df['Year'], df['Digital Marketing'], label = 'Digital Marketing', color = 'black', linewidth = 2.5, linestyle = 'dashdot', marker = 's')

plt.legend()
```

### Adding Grid to the 2D Line Plot using `plt.grid()`

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 Years')

plt.plot(df['Year'], df['Programming'], label = 'Programming', color = '#11b86f', linewidth = 3.5, linestyle = 'dashed', marker = 'o', markersize = 7)
plt.plot(df['Year'], df['Digital Marketing'], label = 'Digital Marketing', color = 'black', linewidth = 2.5, linestyle = 'dashdot', marker = 's')

plt.legend()
plt.grid() # adds grid to the plot/graph

plt.show() # it is mandatory to run after plt.plot() to show the graph/plot outside Jupyter notebook
```

## Scatter Plots

### Loading dataset

```python
students = pd.read_csv('student_IQdata.csv')

students
```

### Feature & Target Scatter Plot using `plt.scatter()`

#### Positive Correlation Scatter Plot

```python
# Relation between Study Hours & IQ Score
plt.xlabel('Study Hours')
plt.ylabel('IQ Score')
plt.title('Relation between Study Hours and IQ Score')

plt.scatter(students['Study_Hour'], students['IQ_Score'], color = 'green')
```

#### No Correlation Scatter Plot

```python
# Relation between Shoe Size & IQ Score
plt.xlabel('Shoe Size')
plt.ylabel('IQ Score')
plt.title('Relation between Shoe Size and IQ Score')

plt.scatter(students['Shoe_Size'], students['IQ_Score'])
```

#### Negative Correlation Scatter Plot

```python
# Relation between Chilling Hours & IQ Score
plt.xlabel('Chilling Hours')
plt.ylabel('IQ Score')
plt.title('Relation between Chilling Hours and IQ Score')

plt.scatter(students['Chilling_Hours'], students['IQ_Score'], color = 'red')
```
