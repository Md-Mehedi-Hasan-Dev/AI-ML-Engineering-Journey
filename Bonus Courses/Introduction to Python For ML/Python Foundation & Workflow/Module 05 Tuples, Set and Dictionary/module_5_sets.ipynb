```python
# Declaration
A = {1, 2, 3}

S = {1, 2, 3, 4, 1, 2} # duplicates will be ignored

print(type(A), A)

print(S)
```

```python
# Empty Structures - empty set declaration
B = [] # empty list

print(type(B))

C = () # empty tuple

print(type(C))

D = {} # not empty set, it is dict

print(type(D))

E = set() # empty set

print(type(E))
```

```python
# Accessing & Checking presence of an element in a set
if (3 in S):
    print("3 is present")
else:
    print("not present")
```

```python
# Accessing & Iterating through all the elements in a set
for element in S:
    print(element, end = " ")
```

## Set Methods

```python
# Mutability of set
s = {5, 1, 2, 4, 3}

print(s)

s.add(10) # insets element in a set (sorted)
s.add(0)

print(s)

s.pop() # pops from the beginning of a set (pops smallest value)
s.pop()

print(s)

s.remove(5) # removes specific element in a set

print(s)
```

## Set Mathematical Methods

```python
# Union & Intersection of sets
a = {1, 2, 3}
b = {2, 3, 4}

print(a.union(b)) # uinon() -> unions two sets

print(a.intersection(b)) # intersection() -> intersects two sets
```

```python
# Disjoint
print(a.isdisjoint(b)) # not disjoint -> there's something common

x = {1, 2, 3}
y = {5, 6, 4}

print(x.isdisjoint(y)) # disjoint -> no common
```

```python
# Subset
p = {1, 2, 3}
q = {5, 6, 4, 1, 2, 3}

print(p.issubset(q))
```
