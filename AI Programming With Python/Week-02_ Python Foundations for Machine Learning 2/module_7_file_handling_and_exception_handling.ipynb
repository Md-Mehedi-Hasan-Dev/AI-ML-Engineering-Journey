## **File Handling & Exception Handling**

## File Read

```python
# reading a file's data and outputing it
file = open("./sample_data/sample.txt", "r")

content = file.read()

print(type(content))

print(content)
```

```python
# reading a file's data line by line
file = open("./sample_data/sample.txt", "r")

content = file.readlines() # outputs as list

print(type(content))

print(content)
```

```python
# Closing files after read/write
print(file.closed) # checking before closing file

file.close() # file closed (always close file other data leakage or other error could occur)

print(file.closed) # checking after closing file
```

### file handling using (with) keyword

```python
# with keyword by defauld handles file closing
with open("./sample_data/sample.txt", "r") as file:
  content = file.readlines()
  print(content)

print(file.closed)
```

```python
# printing lines using loop
with open("./sample_data/sample.txt", "r") as file:
  for line in file:
    print(line) # line contains "\n" and print() adds "\n" at the end so there would be two new lines
```

```python
# removing new line using end in print()
with open("./sample_data/sample.txt", "r") as file:
  for line in file:
    print(line, end="")
```

```python
# removing new line using strip()
with open("./sample_data/sample.txt", "r") as file:
  for line in file:
    l = line.strip() # removes new line and stores in a variable
    print(l)
```

```python
# Data preprocessing
# using strip() to remove new line and filter to remove blank string
with open("./sample_data/sample.txt", "r") as file:
  content = file.readlines()
  content = list(map(str.strip, content))
  filter_content = list(filter(lambda x : len(x) >= 1, content))
  print(filter_content)
```

## File Pointer

### tell()

```python
with open("./sample_data/sample.txt", "r") as file:
  print(file.tell()) # current position (index) of the cursor -> it's at the beginning

  print(file.read())

  print(file.tell()) # the cursor is at the end of the file after reading the file

  print(file.read()) # will read nothing because the cursor is at the end and after this there is no content
```

```python
with open("./sample_data/sample2.txt", "r") as file:
  print(file.tell()) # cursor at 0

  print(file.read(5)) # reads from 0 to 5 position characters

  print(file.tell())

  print(file.read()) # reads rest of the content of the file from the cursors last position

  print(file.tell())
```

### seek()

```python
# seek() -> increment/decrement cursor position (index)
with open("./sample_data/sample.txt", "r") as file:
  print(file.tell())

  file.seek(5)

  print(file.tell())
```

```python
# moving cursor position (index) forward and backward using seek()
with open("./sample_data/sample.txt", "r") as file:
  print(file.tell())

  print(file.read(5))

  print(file.tell())

  file.seek(0) # moving the cursor to the starting position

  print(file.tell())

  print(file.read(5))
```

## Practice Problem on File

### Practice 1 :

Get the text from sample.txt , and count the following


Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce

with open("./sample_data/sample3.txt","r") as file:
  string_list = file.readlines()

  # total line koita
  total_lines = len(string_list)

  # word ber korbo
  number_of_words = list(map(lambda x : len(x.split()),string_list))
  #total words
  total_number_of_words = reduce(lambda x,y : x+y ,number_of_words)

  #cleaning process

  # 1. new line delete
  string_list = list(map(str.strip,string_list))

  #2. space delete

  string_list = list(map(lambda x : x.replace(" ",""),string_list))

  # koita character

  number_of_character = list(map(lambda x : len(x),string_list))

  # total character
  total_number_of_characters = reduce(lambda x,y : x+y ,number_of_character)



with open("./sample_data/counter_of_string.txt","w") as file:
  file.write(f"total line: {total_lines}\ntotal number of words: {total_number_of_words}\ntotal number of char: {total_number_of_characters}")
```

## Exception Handling

```python
n = int(input())

try:
  a = 10 / n
except ZeroDivisionError:
  print("You can't divide a number by 0")

print("Hello World!")
```

```python
try:
  x = y
except ZeroDivisionError:
  print("You can't divide a number by 0")
except NameError as ne:
  print(ne)

print("Hello World!")
```

```python
# exception handling using else & finally keyword
# model training -> file loading error handling
try:
  file = open("./sample_data/data.txt", "r")
except Exception as e:
  print(e)
else:
  # if file exists then execute else block
  print(file.read())
finally:
  print("GPU is stopped")
```
