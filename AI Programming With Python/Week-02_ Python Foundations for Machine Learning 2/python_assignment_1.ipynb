## Answer to the Question No. 1

```python
inp = input()

numbers = [int(num) for num in inp.split()]

print([x for x in numbers if x % 4 == 0 and x % 8 != 0])
```

## Answer to the Question No. 2

```python
def compute_ratio(x, y):
  try:
    return (x + y) / (x * y)
  except ZeroDivisionError:
    return "Error: Division not possible"

inp = input()
numbers = [int(num) for num in inp.split()]
print(compute_ratio(numbers[0], numbers[1]))
```

## Answer to the Question No. 3

```python
with open("./sample_data/data.txt", "r") as file:
  content = file.read()
  maxStr = ""
  for word in content.split():
    if len(maxStr) < len(word):
      maxStr = word
  print(maxStr)
```
