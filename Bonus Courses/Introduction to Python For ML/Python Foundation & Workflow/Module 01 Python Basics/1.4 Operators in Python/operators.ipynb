## Arithmatic Operators

```python
x = 10; y = 3

sum = x + y # summation
sub = x - y # subtraction
mul = x * y # multiplication
div = round(x / y, 2) # divition
rem = x % y # remainder (mod)

print(sum, sub, mul, div, rem)
```

## Comparison Operators

```python
x = 10; y = 3

greater_than = x > y

greater_than_equal = x >= y

print(greater_than, greater_than_equal)

less_than = x < y

less_than_equal = x <= y

print(less_than, less_than_equal)

equal = x == y

not_equal = x != y

print(equal, not_equal)
```
