```python
# list (declaration)
numbers = [100, 20, 30, 40, 50, 100, 100] # int

float_numbers = [10.20, 3.15, 2.4, 3.7, 5.5] # float

fruits = ['apple', 'orange', 'lichi'] # string

mix = [10, 20.3, 'apple'] # mixed

list_within_a_list = [[100, 20, 30], [14, 12, 31], [13, 12, 12]] # 2D list

print(type(numbers))

print(numbers)

print(float_numbers)

print(type(float_numbers[0]))

print(fruits)

print(type(fruits[0]))

print(mix)

print(type(mix[0]), type(mix[1]), type(mix[2]))

print(list_within_a_list)

print(list_within_a_list[0])

print(list_within_a_list[0][2])
```

```python
# accessing elements using index
print(numbers[2])

numbers[2] = 90 # modification

print(numbers[2])
```

```python
# splitting list
new_list = numbers[0:3]

print(new_list)
```
