```python
even = []

# Adding only even numbers to the list (naive way)
for i in range(101):
    if i % 2 == 0:
        even.append(i)

print(even)
```

## List Comprehension

```python
random_list = [x for x in range(1, 101) if x % 2 == 0]

print(random_list)
```

```python
random_list2 = [x*2 for x in range(1, 101) if x % 2 != 0]

print(random_list2)
```

## List Comprehension with List of String

```python
# Without List Comprehension (naive way)
fruits = ['apple', 'orange', 'lichi']

capitalize_fruits = []

upper_fruits = []

for fruit in fruits:
    capitalize_fruits.append(fruit.capitalize())
    upper_fruits.append(fruit.upper())

print(capitalize_fruits)

print(upper_fruits)
```

```python
# With List Comprehension
lower_fruits = [fruit.lower() for fruit in upper_fruits]

print(lower_fruits)
```
