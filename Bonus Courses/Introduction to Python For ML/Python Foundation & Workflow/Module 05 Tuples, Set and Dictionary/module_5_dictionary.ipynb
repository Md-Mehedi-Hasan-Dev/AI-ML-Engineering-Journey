## Dictionaries in Python

```python
# Empty
dic = {}

print(type(dic), dic)

# declaration
dic = {"name" : "Mehedi", "age" : 23, "address" : "Dhaka", "numbers" : [10, 20, 30]}

print(type(dic), dic)
```

```python
# Accessing value using key

# dic_name[key]
print(dic["name"])
print(dic["address"])
print(dic["numbers"])

# using get()
print(dic.get("age"))
```

```python
# Mutability
dic["name"] = "Mehedi Hasan"

print(dic["name"])

print("age :", dic["age"])

dic = {"name" : "Mehedi", "age" : 23, "address" : "Dhaka", "numbers" : [10, 20, 30], "age" : 30}

print(dic)
```

## Dictionary Methods

```python
# printing default values for keys not in a dictionary
print(dic)

print(dic.get("math_marks")) # this key is not available

print(dic.get("math_marks", 0)) # because this key is not in the dic we are giving a default value if not available
```

```python
# adding a value to a dictionary
print(dic)

dic["math_marks"] = 95

print(dic)

# using update()
dic.update({"english_marks" : 90})

print(dic)
```

```python
# deleting key in dictionary
del dic["english_marks"]

print(dic)
```

```python
# copying a dictionary to another dictionary
dic_2 = dic.copy()

print(dic_2)
```

```python
# Key immutability
# dic_3 = {[10, 20, 30] : "a"} # will give error because list is mutable and we can't use this for key
# dic_3 = {{"name" : "adil", "age": 20} : 65} # error - dictionary is mutable so can't use it as key
dic_3 = {(10, 20, 30) : "a", "key" : [10, 20, 30], "k" : (10, 20, 30), "ky" : {10, 20, 30}, "keyyy" : {"name" : "adil", "age": 20}}

print(dic_3)
```

## Dictionary View Methods - iterating over dictionary

```python
# keys()
dic_4 = {"name" : "Mehedi", "age" : 23, "address" : "Dhaka", "number" : [10, 2, 30], "age" : 30}

keys = dic_4.keys()

print(keys)

dic_4["math_marks"] = 99

print(keys) # real time update no need to call keys() everytime
```

```python
# values()
values = dic_4.values()

print(values)

dic_4["math_marks"] = 100

print(values) # real time update
```

```python
# items() -> returns key-value pairs as tuples and the whole dictionary in a list
items = dic_4.items()

print(items) # real time update
```

```python
# iterating a dictionary using item()
for key,value in dic_4.items():
    print(key, value)
```

## Dictionary Comprehension

```python
square = {x:x**2 for x in range(1, 11)} # x**2 = x^2

print(square)
```

```python
square = {x:x**2 for x in range(1, 11) if x%2==0}

print(square)
```

```python
# combining two lists (one as key other as value) into a dictionary using zip()
co_ordinates = [(10, 10.5), (20.5, 192), (101, 102)]
locations = ["dhaka", "chattogram", "sylhet"]

exact_location = {co_or : loc for co_or, loc in zip(co_ordinates, locations)}

print(exact_location)
```
