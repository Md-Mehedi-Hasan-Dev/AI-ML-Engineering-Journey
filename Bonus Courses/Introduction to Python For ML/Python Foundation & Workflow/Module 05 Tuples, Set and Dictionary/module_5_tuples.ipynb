## Tuples

```python
# Declaration
tup = (10, 20, 30, 40)

float_tup = (10.5, 11.2)

mixed_tup = (10, 12.5, "str", True)

print(type(tup), tup)

print(type(float_tup), float_tup)

print(type(mixed_tup), mixed_tup)
```

```python
# Important Issue - How to declare one element in a tuple
tup1 = (10) # not tuple type, it is int type becase only one element

print(type(tup1))

tup2 = (10, ) # it is a way to declare a one element tuple

print(type(tup2))

tup3 = tuple([10]) # another way to declare a one element tuple

print(type(tup3))
```

```python
# Converting list to tuple using tuple()
lst1 = [10, 20, 30, 40]

tup4 = tuple(lst1)

print(type(tup4), tup4)
```

```python
# Converting tuple to list using list()
tup5 = (11, 12, 13)

lst2 = list(tup5)

print(type(lst2), lst2)
```

```python
# Accessing elements in tuple using indexing
tup6 = (1, 20, 30, 10.5, "str", False)

print(tup6[4])
```

```python
# Slicing in tuple
new_tup = tup6[0:3]

print(new_tup)
```

```python
# mutability - list
lst3 = [10, 20, 30, 40, 50]
lst3.append(60)
lst3[0] = 100
print(lst3)

# immutability
tup7 = (10, 20, 30, 40)
#tup[1] = 100 # not possible - error
print(tup7)
```

## Tuple Methods

```python
# count()
tup8 = (10, 20, 30, 40, 10, 20, 30)

print(tup8.count(10))
```

```python
# index()
print(tup8.index(10))

print(tup8.index(20))

print(tup8.index(40))
```
