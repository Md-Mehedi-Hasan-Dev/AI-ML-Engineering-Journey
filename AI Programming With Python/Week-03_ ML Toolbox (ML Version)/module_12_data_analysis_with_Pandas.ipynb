## DataFrame from In-built Data Structure Datas

### Converting list to DataFrame

```python
# 2D list to DataFrame
my_list = [
    ['Alice', 25],
    ['Bob', 30],
    ['Charlie', 28]
]


list_df = pd.DataFrame(my_list, columns = ['Name', 'Age'], index = [1, 2, 3])

print(type(list_df))

list_df
```

### Converting tupele to DataFrame

```python
# 2D tuple to DataFrame
my_tuple = (
    ('Alice', 25),
    ('Bob', 30),
    ('Charlie', 28)
)

tuple_df = pd.DataFrame(my_tuple, columns = ['Name', 'Age'])

print(type(tuple_df))

tuple_df
```

### Converting dictionary to DataFrame

```python
# 2D dictionary to DataFrame
my_dict = {
    'Name' : ['Alice', 'Bob', 'Charlie'],
    'Age' : [25, 30, 28]
}

dict_df = pd.DataFrame(my_dict) # key of the dictionary is taken as column labels

print(type(dict_df))

dict_df
```

### Converting list of dictionaries (Data in JSON like structure) to DataFrame

```python
# list of dictionaries to DataFrame
my_data = [
    {'Name' : 'Alice', 'Age' : 25, 'City' : 'New York'},
    {'Name' : 'Bob', 'Age' : 30, 'City' : 'Paris'},
    {'Name' : 'Charlie', 'Age' : 28}
]

data_df = pd.DataFrame(my_data)

print(type(data_df))

data_df
```

### Conventional way of accessing row & col in Pandas

#### .loc[row (index), col (feature)] -> accessing row & col by their labels or via a boolean array

```python
# .loc[row_start : row_end, col_start : col_end]

# row accessing
print(df.loc[0])

print(type(df.loc[0]))
```

##### Accessing multiple rows using list (the rows want to access)

```python
df.loc[[2, 3, 19]]
```

##### Accessing multiple rows using range (the range rows want to access)

```python
# .loc slicing are inclusive of both the start and stop labels
df.loc[3 : 7]
```

##### Accessing single col

```python
df.loc[:, 'Python Marks']
```

##### Accessing multiple cols using list (the col want to access) -> DataFrame

```python
print(df.loc[:, ['Python Marks', 'Algorithm Marks']])

print(type(df.loc[:, ['Python Marks', 'Algorithm Marks']]))
```

##### Accessing rows with cols

```python
df.loc[3:7, ['Python Marks', 'Algorithm Marks', 'CompletionStatus']]
```

## Changing Index & Columns and .iloc

### Changing Index (row labels) using .set_index() -> converts one or more existing columns into the DataFrame's row labels (index)

```python
# inplace = True -> parameter modifies the original DataFrame directly
df_index = df.set_index('StudentID')

df_index
```

### Accessing rows index with cols using .iloc -> used for integer-location based indexing where the index is not numberic

```python
# .iloc slicing are exclusive of the stop index
df_index.iloc[3 : 7]
```

### Accessing cols with rows using .iloc without numeric indexing

```python
df_index.iloc[:, 0:4]
```

### Changing Cols labels using .rename() -> used to alter the index (row labels) or column labels of a DataFrame by passing a dictionary as parameter or  a dictionarry as positional argument

```python
# inplace = True -> parameter modifies the original DataFrame directly
df_new = df.rename(columns = {'FullName' : 'Full Name', 'Algorithm Marks' : 'Algo Marks'})

df_new
```

## Dropping rows & cols, Assigning/changing values & Iteration on a DataFrame

```python
df
```

### Dropping rows or cols using .drop(index, columns, inplace) -> used to remove rows or columns from a DataFrame by specifying their corresponding labels

#### Dropping rows

```python
# inplace = True -> parameter modifies the original DataFrame directly
print(df.drop(0)) # returns the copy of the DataFrame if inplace = True is not given
```

#### Dropping cols

```python
# inplace = True -> parameter modifies the original DataFrame directly
# To remove columns, pass the column name(s) and specify axis=1, or use the explicit columns parameter
print(df.drop('Instructor', axis = 1))
```

### Accessing/Assigning/Changing the values of DataFrame using .loc

```python
df.loc[1, 'Python Marks'] = 90

df.loc[1, 'CompletionStatus'] = 'Completed'

df.head()
```

```python
# adding 2 marks to Python Marks in first 5 students
df.loc[0:4, 'Python Marks'] += 2

df.head()
```

### Iterating DataFrame

#### .iterrows() -> is a generator function used to iterate over the rows of a DataFrame as (index, Series) pairs

```python
for i, series in df.iterrows():
    print(f"{i} : {series}") # prints index : rows as series
```

#### .itertuples ->  iterates over DataFrame rows as Python namedtuple objects

```python
# .itertuples significantly faster and more memory-efficient than df.iterrows()
# if we don't want to show index for the tuples we can set index = False in method parameter
for i in df.itertuples(index = False):
    print(i)
```

## Sorting DataFrame

```python
df
```

### using .sort_values(column labels (by = single or [list]), ascending (boolean -> single or [list for list passed in by]), inplace)

#### sorting based on column labels

```python
# inplace = True -> parameter modifies the original DataFrame directly
# na_position (parameter) -> by default NaN are placed at the end can change it to front by na_position = 'first'
# ascending (parameter) -> by default True
print(df.sort_values('Data Structure Marks', ascending = False))
```

#### Sorting based on column labels and passing list of labels

```python
# if some 'Data Structure Marks' have same values than they will be sorted by 'Python Marks'
print(df.sort_values(['Data Structure Marks', 'Python Marks'], ascending = [False, True]))
```

## Filtering Data based on Condition using .loc

```python
# datas of not started course
not_started = df.loc[df['CompletionStatus'] == 'Not Started']

not_started
```

```python
# datas of completed course
comleted = df.loc[df['CompletionStatus'] == 'Completed']

comleted
```

```python
# datas that completed and 'Data Structure Marks' greater or equal to 90
star_student = df.loc[(df['CompletionStatus'] == 'Completed') & (df['Data Structure Marks'] >= 90)] # use bracket for every condition

star_student
```
