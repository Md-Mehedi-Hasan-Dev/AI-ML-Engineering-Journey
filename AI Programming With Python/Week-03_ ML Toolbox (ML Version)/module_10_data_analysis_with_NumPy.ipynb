## Ndarray

```python
import numpy as np

# 1d array declaration
arr1 = np.array([1, 2, 3, 4, 5]) # we can pass any iterable in the method

print(type(arr1))
print(arr1.ndim) # printing dimension of the iterable
# print(arr1)

# 2d array declaration -> alot of 1d arrays in another array
arr2 = np.array([   [1, 2, 3], # row1
                    [4, 5, 6] # row2
                ])
print(arr2.ndim)
# print(arr2)

# 3d array declaration -> alot of 2d arrays in another array -> mainly used in Image Processing
arr3 = np.array([
    # 1st floor
    [   [1, 2, 3], # row1
        [4, 5, 6] # row2
    ],
    # 2nd floor
    [   [1, 2, 3], # row1
        [4, 5, 6] # row2
    ],
    # 3rd floor
    [   [1, 2, 3], # row1
        [4, 5, 6] # row2
    ]
])

print(arr3.ndim)
# print(arr3)
```

### slicing

```python
# 1d array slicing = arr[start : end : step]
print(arr1)

arr1_mod = arr1[1:4]

print(arr1_mod) # excludes the last one
```

```python
arr = np.array([1, 2, 3, 4, 5])
arr[1:4] = 0
print(arr)
```

```python
# changing sliced portion changes original array as slice portion is a pointer to the actual array
# slice = (view of the actual array)
print(arr1)

arr1_mod[0] = 50 # chaged a value in the slice

print(arr1_mod)
print(arr1) # arr1 value changed
```

```python
# making a copy of the array as a slice using .copy() -> changing in slice won't change original array
print(arr1)

arr1_mod = arr1[1:4].copy()

arr1_mod[2] = 200 # changed a value in the slice

print(arr1_mod)
print(arr1) # arr1 hasn't changed
```

```python
# 2d array slicing = arr[row_start : row_end : row_step , col_start : col_end : col_step]
print(arr2)

# getting a row
row_0 = arr2[0:1, ]
print(row_0)

row_1 = arr2[1: , ]
print(row_1)

# getting a col
col_0 = arr2[ : , 0:1]
print(col_0)

col_1 = arr2[ : , 1:2]
print(col_1)

# getting a portion
portion = arr2[ : , 0:2]
print(portion)
```
