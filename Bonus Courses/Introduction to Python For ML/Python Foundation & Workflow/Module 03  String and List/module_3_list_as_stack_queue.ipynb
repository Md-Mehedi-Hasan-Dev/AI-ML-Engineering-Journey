## List as Stack

```python
stack = [] # List as a stack

print(type(stack))

print(len(stack))
```

```python
# Push/insert elements in stack
stack.append(1)
stack.append(2)
stack.append(3)
stack.append(4)
stack.append(5)

print(stack)
```

```python
# Pop/remove elements in stack
print(stack.pop())
```

```python
# Accessing last element in stack
print(f"Top element of Stack: {stack[-1]}")
```

## List as Queue

```python
queue = []

# pushing/insertion elements in a queue
queue.append(1)
queue.append(2)
queue.append(3)
queue.append(4)
queue.append(5)

print(queue)
```

```python
# pop/remove elements in a queue
print(queue.pop(0))

print(f"Front of Queue: {queue[0]}")
```
