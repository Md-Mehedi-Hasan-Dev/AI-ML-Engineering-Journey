## Problem - Even, Odd, Positive and Negative

```python
inp = input()

inp_value = input()

numbers = [int(num) for num in inp_value.split()]

pos = 0; neg = 0; even = 0; odd = 0

for n in numbers:
    if n > 0:
        pos += 1
    elif n < 0:
        neg += 1

    if n % 2 == 0:
        even += 1
    else:
        odd += 1

print(f"Even: {even}")

print(f"Odd: {odd}")

print(f"Positive: {pos}")

print(f"Negative: {neg}")
```
